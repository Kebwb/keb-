```python
from os import remove

from keyring.backends.Windows import missing_deps
%matplotlib inline
import pandas as pd
import numpy as np #数字和矩阵
import sklearn.linear_model #提供机器学习算法
import matplotlib as mpl #画图库
import matplotlib.pyplot as plt

mpl.rc('axes', labelsize=14) #设置字号大小
mpl.rc('xtick', labelsize=12)  #x轴
mpl.rc('ytick', labelsize=12)  #y轴
# 加载数据
oecd_bli = pd.read_csv("./ml_dataset/oecd_bli_2015.csv", thousands=',')
gdp_per_capita = pd.read_csv("ml_dataset/gdp_per_capita.csv", thousands=',', delimiter='\t', encoding='latin1', na_values="n/a")
```

```python
oecd_bli = oecd_bli[oecd_bli["INEQUALITY"]=="TOT"] #筛选TOT
oecd_bli.head()
```

```python
oecd_bli = oecd_bli.pivot(index="Country", columns="Indicator", values="Value")
oecd_bli.head()#设置行名和列名列出
```

```python
gdp_per_capita.info()#查看表基本信息
```

```python
gdp_per_capita.rename(columns={"2015": "GDP per capita"}, inplace=True)
```

```python
gdp_per_capita.set_index("Country", inplace=True)
```

```python
gdp_per_capita.head()
```

```python
full_country_stats = pd.merge(left=oecd_bli, right=gdp_per_capita, left_index=True, right_index=True)
full_country_stats.head() #两张表按国家名做连接，生成一张宽表
```

```python
full_country_stats.sort_values(by="GDP per capita", inplace=True)
full_country_stats.head()
```

```python
full_country_stats.info()
```

```python
country_stats=full_country_stats[["GDP per capita",'Life satisfaction']]
country_stats.head()
```

```python
country_stats.plot(kind='scatter', x="GDP per capita", y='Life satisfaction')
plt.show()
```
