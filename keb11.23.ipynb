```python
%matplotlib inline

import pandas as pd
import numpy as np
import sklearn.linear_model
import matplotlib as mpl
import matplotlib.pyplot as plt

mpl.rc('axes', labelsize=14)
mpl.rc('xtick', labelsize=12)
mpl.rc('ytick', labelsize=12)
# 加载数据
oecd_bli = pd.read_csv("./ml_dataset/oecd_bli_2015.csv", thousands=',')
gdp_per_capita = pd.read_csv("./ml_dataset/gdp_per_capita.csv", thousands=',', delimiter='\t',encoding='latin1', na_values="n/a")
```

```python
oecd_bli.head()  #显示前五行
```

```python
oecd_bli.info()  #查看基本信息
```

```python
oecd_bli["Inequality"].value_counts() #查看元素个数
```

```python
oecd_bli["INEQUALITY"].value_counts()
```

```python
oecd_bli = oecd_bli[oecd_bli["INEQUALITY"]=="TOT"] #筛选TOT
oecd_bli.head()
```

```python
oecd_bli = oecd_bli.pivot(index="Country", columns="Indicator", values="Value")
oecd_bli.head()#设置行名和列名列出
```

```python
gdp_per_capita.info()
```

```python
gdp_per_capita.head
```

```python
gdp_per_capita.rename(columns={"2015": "GDP_per_capita"}, inplace=True)
```

```python
gdp_per_capita.set_index("Country", inplace=True)
```

```python
gdp_per_capita.head()
```

```python
full_country_stats = pd.merge(left=oecd_bli, right=gdp_per_capita, left_index=True, right_index=True)
full_country_stats.head()
```

```python
full_country_stats.sort_values(by="GDP_per_capita", inplace=True)
full_country_stats.head()
```

```python
full_country_stats.info()
```

```python
country_stats=full_country_stats[["GDP_per_capita",'Life expectancy']]
country_stats.head()
```

```python
country_stats.plot(kind='scatter', x="GDP_per_capita", y='Life expectancy')
plt.show()
```

```python
country_stats.count()
```

```python
country_stats. to_csv('country_stats.csv')
```
